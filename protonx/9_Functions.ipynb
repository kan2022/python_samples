Phạm Vi - Scoope

```python
def myfunc():
  x = 300
  print(x)

myfunc() # 300

print(x)
```

```python
def myfunc():
  x = 300
  def myinnerfunc():
    print(x)
  myinnerfunc()

myfunc() # 300
```

```python
def myfunc():
  x = 300

print(x)
```

```python
def myfunc():
  global x
  x = 300
print(x)
```
