```python
def exercise9():

    # Lập trình tại đây

    name = 'Nguyễn văn X'
    age = 22
    company = 'X Company'
    position = 'Software Engineer'

    return name, age, company, position


# TEST CODE
try:
    name, age, company, position = exercise9()
    print("Giá trị name đã được khai báo đúng?: {}".format(type(name).__name__ == 'str' and name == 'Nguyễn văn X'))
    print("Giá trị age đã được khai báo đúng?: {}".format(type(age).__name__ == 'int' and age == 22))
    print("Giá trị company đã được khai báo đúng?: {}".format(type(company).__name__ == 'str' and company == 'X Company'))
    print("Giá trị position đã được khai báo đúng?: {}".format(type(position).__name__ == 'str' and position == 'Software Engineer'))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise3():

    # Lập trình tại đây

    a = -1
    b = 10

    return a, b


# TEST CODE
try:
    a, b = exercise3()
    print("Giá trị của a và b đã được khai báo đúng?: {}".format(a == -1 and b == 10))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise4():

    # Lập trình tại đây

    a = 2
    b = 5
    c = a+b

    return a, b, c


# TEST CODE
try:
    a, b, c = exercise4()
    print("Giá trị của a đã được khai báo đúng?: {}".format(a == 2))
    print("Giá trị của b đã được khai báo đúng?: {}".format(b == 5))
    print("Giá trị của c đã được tính toán đúng?: {}".format(c == 7))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise8():

    # Lập trình tại đây

    job = "abcxyz"
    out = "Tôi làm {}".format(job)

    return job, out


# TEST CODE
try:
    job, out = exercise8()
    print("Giá trị job đã được khai báo đúng?: {}".format(len(job) > 0))
    print("Giá trị out đã được tính toán đúng?: {}".format("Tôi làm " in out and out[len("Tôi làm "):] == job))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise10():

    # Lập trình tại đây

    print('Tôi là vua của phép cộng')
    a = int(input('Vui lòng nhâp a: '))
    b = int(input('Vui lòng nhâp b: '))
    c = a+b
    print('a + b = {}'.format(c))

    return a, b, c


# TEST CODE
from unittest.mock import patch
try:
    @patch('builtins.input', side_effect=['3', '5'])
    def test(self):
        return exercise10()
    a, b, c = test()
    print("Hàm nhập giá trị a thực hiện đúng?: {}".format(a == 3))
    print("Hàm nhập giá trị a thực hiện đúng?: {}".format(b == 5))
    print("Hàm tính tổng a và b thực hiện đúng?: {}".format(c == 8))
except Exception as e:
    print("Lỗi thực thi: ", e)
```
