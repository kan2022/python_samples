```python
def exercise1():

    # Lập trình tại đây
    a = True
    b = False
    c = (a or b) and (b or a) or b

    return a, b, c


# TEST CODE
try:
    a, b, c = exercise1()
    print("Giá trị của a đã được khai báo đúng?: {}".format(a))
    print("Giá trị của b đã được khai báo đúng?: {}".format(not b))
    print("Giá trị của c đã được tính toán đúng?: {}".format(1 == 1))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise2(x):

    # Lập trình tại đây
    msg = None

    if 10 <= x <= 20:
        msg = "Nằm trong khoảng [10, 20]"
    else:
        msg = "Không nằm trong khoảng [10, 20]"

    return msg


# TEST CODE
try:
    msg1 = exercise2(3)
    msg2 = exercise2(20)
    print("Điều kiện 1 được lập trình: {}".format(msg1 == "Không nằm trong khoảng [10, 20]"))
    print("Điều kiện 2 được lập trình: {}".format(msg2 == "Nằm trong khoảng [10, 20]"))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise3(age, at_hanoi):

    # Lập trình tại đây
    msg = None

    if age <= 50 and at_hanoi:
        msg = "Hoàn toàn đủ điều kiện"
    else:
        msg = "Không đủ điều kiện"

    return msg


# TEST CODE
try:
    msg1 = exercise3(50, False) # 50 tuổi và không ở Hà Nội
    msg2 = exercise3(20, False) # 20 tuổi và không ở Hà Nội
    msg3 = exercise3(10, True) # 10 tuổi và ở Hà Nội
    msg4 = exercise3(60, True) # 60 tuổi và ở Hà Nội
    print("Trường hợp đầu tiên có kết quả: {}".format(msg1 == "Không đủ điều kiện"))
    print("Trường hợp thứ hai có kết quả: {}".format(msg2 == "Không đủ điều kiện"))
    print("Trường hợp thứ ba có kết quả: {}".format(msg3 == "Hoàn toàn đủ điều kiện"))
    print("Trường hợp thứ tư có kết quả: {}".format(msg4 == "Không đủ điều kiện"))
except Exception as e:
    print("Lỗi thực thi: ", e)
```
