```python
class Oto():
  def __init__(self, color, engine):
    self.num_wheels = 4
    self.__engine = engine
    self.color = color

  def run(self):
    print("Vroom")

  def get_info(self):
    print("the color", self.color)

  def __get_engine_info(self):
    print("the engine: ", self.__engine)

  def get_engine_info(self):
    self.__get_engine_info()
```

```python
class HondaOto(Oto):
  def __init__(self, color, engine, is_mui_tran):
    super().__init__(color, engine)
    self.is_mui_tran = is_mui_tran

  def run(self):
    print("Bạch bạch bạch")
```

```python
hondaoto1 = HondaOto('Đỏ', 'v12', True)
hondaoto1.num_wheels
hondaoto1.run()
```

```python
hondaoto1.num_wheels
```

```python
hondaoto1.is_mui_tran
```
