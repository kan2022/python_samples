```python
def exercise1(a, b):

    # Lập trình tại đây
    c = 0
    if a > b:
        print("a phải nhỏ hơn hoặc bằng b")
        return 0
    else:
        for i in range(a, b+1):
            if i % 2 == 1:
                c += i

    return c


# TEST CODE
from unittest.mock import patch
from io import StringIO
try:
    @patch('sys.stdout', new_callable=StringIO)
    def test1(mock_stdout):
        exercise1(20, 1)
        t1 = mock_stdout.getvalue().strip() == "a phải nhỏ hơn hoặc bằng b"
        return t1

    def test2():
        c = exercise1(3, 13)
        return c == 48

    def test3():
        c = exercise1(-3000, 3000)
        return c == 0

    def test4():
        c = exercise1(-10000, 5000)
        return c == -18750000

    t1 = test1()
    t2 = test2()
    t3 = test3()
    t4 = test4()
    print("Mệnh đề đầu tiên thực hiện đúng?: {}".format(t1))
    print("Mệnh đề thứ hai thực hiện đúng?: {}".format(t2 and t3 and t4))
except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise2(n):

    # Lập trình tại đây
    x = None
    if n < 0:
        print('n phải lớn hơn hoặc bằng 0')
        x = -1
    elif n == 0:
        x = 0
    elif n == 1:
      x = 1
    elif n == 2:
      x = 1
    else:
        x = 0
        y = 1
        z = 1
        for i in range(n-2):
          x = y + z
          z = y
          y = x
    return x

# TEST CODE
from unittest.mock import patch
from io import StringIO
try:
    @patch('sys.stdout', new_callable=StringIO)
    def test1(mock_stdout):
        exercise2(-1)
        t1 = mock_stdout.getvalue().strip() == "n phải lớn hơn hoặc bằng 0"
        return t1

    def test2():
        x = exercise2(100)
        t2 = x == 354224848179261915075
        return t2

    def test3():
        x = exercise2(10)
        t3 = x == 55
        return t3

    def test4():
        x = exercise2(0)
        t4 = x == 0
        return t4

    def test5():
        x = exercise2(-10)
        t5 = x == -1
        return t5

    t1 = test1()
    t2 = test2()
    t3 = test3()
    t4 = test4()
    t5 = test5()
    print("Mệnh đề đầu tiên thực hiện đúng?: {}".format(t1 and t5))
    print("Mệnh đề thứ hai thực hiện đúng?: {}".format(t2 and t3 and t4))
except Exception as e:
  print("Lỗi thực thi: ", e)
```
