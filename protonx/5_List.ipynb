```python
from ast import Add
from os import popen
sothu = ['hươu cao cổ', 'cừu', 'nhím', 'cá', 'heo', 'lạc đà', 'bò sữa', 'mèo']
while True:
  print('Chào quản lý vườn thú, Tôi có thể giúp bạn')
  print('0. In danh sách')
  print('1. Truy cập')
  print('2. Thêm thú nuôi')
  print('3. Thay thế các thú nuôi')
  print('4. Chuyển thú nuôi đi')
  op = int(input(''))
  if op == 0:
    print('--------------')
    print('Dánh sách thú nuôi:')
    i = 0
    while i < len(sothu):
      print('{}. {}'.format(i+1, sothu[i]))
      i += 1
    print('--------------')
  elif op == 1:
    print('Nhập vị trí thú nuôi:')
    pos = int(input(''))
    try:
      print('{} đang ở đây'.format(sothu[pos]))
    except:
      print('Không có động vật nào ở đây')
  elif op == 2:
    print('Nhập thêm thú nuôi')
    sothu.append(input(''))
    print('Đã thêm {}'.format(sothu[-1]))
  elif op == 3:
    try:
      print('Nhập vị trí muốn thay thế:')
      pos = int(input(""))
      print('Nhập thú nuôi mới:')
      new = input('')
      old = sothu[pos]
      sothu[pos] = new
      print('Đã thay thế {} thành {}'.format(old, new))
    except ValueError:
      print('Đầu vào có vấn đề.')
    except IndexError:
      print('Không tồn tại vị trí trong mảng')
    except:
      print('Không thể thay đổi')
  elif op == 4:
    try:
      print('Nhập thú nuôi muốn tạm biệt:')
      animal = input('')
      sothu.remove(animal)
      print('Đã tiễn biệt {}'.format(animal))
    except:
      print('Đầu vào có vấn đề.')
  else:
    break
```

```python
def exercise1(aList, i, value):

    # Lập trình tại đây
    try:
        aList[i] = value
    except:
        pass
    return aList


# TEST CODE
FIRST = '1'
try:
    q = FIRST

    inp = [
        (['thịt gà 🐔', 'thịt heo 🐖', 'phở bò 🍲', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅'], 1, 'Sandwich 🥪'),
        (['thịt gà 🐔', 'thịt heo 🐖', 'phở bò 🍲', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅'], 2, 'Sandwich 🥪'),
        (['thịt gà 🐔', 'thịt heo 🐖', 'phở bò 🍲', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅'], 10, 'Sandwich 🥪')
        ]
    out = [
        ['thịt gà 🐔', 'Sandwich 🥪', 'phở bò 🍲', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅'],
        ['thịt gà 🐔', 'thịt heo 🐖', 'Sandwich 🥪', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅'],
        ['thịt gà 🐔', 'thịt heo 🐖', 'phở bò 🍲', 'rau cải 🥬', 'trứng gà 🥚', 'cà chua 🍅']
        ]


    res_list = [False for _ in range(len(inp))]
    for i, inpi in enumerate(inp):
        res_list[i] = exercise1(*inpi) == out[i]

    ans = 'Câu {}:\n'.format(q)
    for i, item in enumerate(res_list):
        if not res_list[i]:
            ans += "\nTest case {} được tính đúng?: {}.\n\t Giá trị đầu vào: {}".format(i + 1, res_list[i], inp[i])
        else:
            ans += "\nTest case {} được tính đúng?: {}.".format(i + 1, res_list[i])
    print(ans)

except Exception as e:
    print("Lỗi thực thi: ", e)
```

```python
def exercise2(aList):

    # Lập trình tại đây
    result = []
    for i in range(len(aList)):
        n = 0
        for j in aList[0:i+1]:
            n += j
        result.append(n)
    return result


# TEST CODE
SECOND = '2'
try:
    q = SECOND
    inp = [
        [1, 2, 3, 4],
        [-1, 2000, 33999, 3],
        [1000, -200, 30, 3, 50]
        ]
    out = [
        [1, 3, 6, 10],
        [-1, 1999, 35998, 36001],
        [1000, 800, 830, 833, 883]
        ]

    res_list = [False for _ in range(len(inp))]
    for i, inpi in enumerate(inp):
        res_list[i] = exercise2(inpi) == out[i]

    ans = 'Câu {}:\n'.format(q)
    for i, item in enumerate(res_list):
        if not res_list[i]:
            ans += "\nTest case {} được tính đúng?: {}.\n\t Giá trị đầu vào: {}".format(i + 1, res_list[i], inp[i])
        else:
            ans += "\nTest case {} được tính đúng?: {}.".format(i + 1, res_list[i])
    print(ans)
except Exception as e:
    print("Lỗi thực thi: ", e)
```
