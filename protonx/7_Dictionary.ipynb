```python
aDict = {
    'go': 'đi',
    'bear': 'gấu',
    'fish': 'cá',
    'whale': 'cá voi',
    'cat': 'mèo'
}
```

```python
while True:
  word = input('Nhập từ muốn dịch: ')
  word = word.strip().lower()
  print(word)
  if word in aDict:
    print(aDict[word])
  else:
    print('Không tồn tại từ này')
  break
```
