Để chứng minh công thức tính tổng dãy số dương từ 1 đến (n), ta sử dụng công thức sau:

S = 1 + 2 + 3 + … + n

Công thức này có thể được chứng minh bằng phương pháp quy nạp hoặc sử dụng một cách tiếp cận thông minh do Gauss tìm ra. Ở đây, tôi sẽ trình bày cách tiếp cận của Gauss:

Giả sử ta có một dãy số từ 1 đến (n), ta viết lại dãy số đó theo thứ tự ngược lại:

S = n + (n - 1) + (n - 2) + … + 1

Khi cộng mỗi cặp số ở hai đầu của dãy số, tức là (1 + n), (2 + (n-1)), và cứ tiếp tục như vậy, ta thấy rằng tổng của mỗi cặp luôn bằng (n + 1). Vì có tổng cộng (n) số, và mỗi số được cộng đôi một lần, ta có thể nhận ra rằng tổng cuối cùng sẽ là:

S = n(n+1) ÷ 2


Đây chính là công thức để tính tổng của dãy số dương từ 1 đến (n), và nó được chứng minh thông qua việc cộng đối xứng các phần tử trong dãy số.

```python
import numpy as np

def exercise1():
    # Định nghĩa x
    x = np.array([1, 2, 3])

    # Định nghĩa y
    y = np.array([-1, -2, 3])

    # Tính chuẩn x và gán vào biến x_norm2
    x_norm2 = 0
    for i in x:
        x_norm2 += i**2
    x_norm2 = np.sqrt(x_norm2)
    # Tính chuẩn y và gán vào biến y_norm2
    y_norm2 = 0
    for i in y:
        y_norm2 += i**2
    y_norm2 = np.sqrt(y_norm2)

    return [round(x_norm2, 2), round(x_norm2, 2)]

x_norm2, y_norm2 = exercise1()
print(round(x_norm2, 2), round(y_norm2,2))
```

# 1.3.2 Hadamard product
Hadamard product sẽ thực hiện phép nhân elementwise để tạo ra vector mới.

```python
import numpy as np
from numpy import dot
from numpy.linalg import norm

def cal_cos(x, y):
    # Lâp trình tại đây
    cos_sim = 0
    cos_sim = np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))
    return round(cos_sim, 4)

x = np.array([1, 2, 3])
y = np.array([2, 3, 4])
z = np.array([-2, -2, 3])

print(cal_cos(x, y))
print(cal_cos(x, z))
```

# 2.5 Matrix multiplication (nhân ma trận) - Dot Product
Cho hai ma trận ***A( m × n ), B( n × p)***, tích của ma trận được ký hiệu là ***C = AB*** khi đó ***C*** có kích thước ***m × p***. Ma trận C được tạo thành được tính như hình dưới đây:

**Chú ý**: để nhân ma trạn số cột của ma trận thứ mhất phải bằng số hàng của ma trận thư hai.
Một số tính chất của ma trận:
- Phép nhân ma trận **không** có tính chất giao hoán.
- Phép nhân ma trận có tính kết hợp: ***ABC = (AB)C = A(BC)***
- Phép nhân ma trận có tính giao hoán: ***A(B + C) = AB + AC***
- Chuyển vị của một tích chuyển vị theo thứ tự ngược lại: (AB)T = B** **T × A** **T

```python
A = np.array([[1, 7], [2, 4]])
B = np.array([[3, 3], [5, 2]])

np.dot(A, B)
```

```python
import numpy as np
def exercise12(A, B):
    # Lập trình tại đây
    result = np.matmul(A.T, A)
    result = np.matmul(result, A)
    PreResult = np.matmul(B, B.T)
    result = np.matmul(result, PreResult)
    return result

A = np.array([[1, -1, 2 ],[2, 3, 4 ],[4, 1 , 5]])
B = np.array([[2, -1],[3, -4],[4, -5]])

print(exercise12(A, B))
```
